```python
from pyspark.sql.functions import explode, split, array_distinct
titles = spark.read.table("db.titles")
from pyspark.sql import functions as F

all_genres = (
    titles
    .select(F.explode(F.split("genre", r",\s*")).alias("genre"))  # split + strip spaces in one go
    .distinct()                                                    # deduplicate across all rows
    .orderBy("genre")
)
all_genres.show()
```
